```python
# import pandas
import pandas as pd

df = pd.read_csv('/content/Diversity in tech companies.csv')

# print header
print("First 5 rows of the DataFrame:")
print(df.head())

print("\nColumn names and their data types:")
print(df.info())

print("\nMissing values per column:")
print(df.isnull().sum())

# assign filtered data frame
df_filtered = df[(df['Year'] >= 2014) & (df['Year'] <= 2018)].copy()

diversity_cols = ['% White', '% Asian', '% Latino', '% Black', '% Multi', '% Other', '% Undeclared']

for col in diversity_cols:
    # Replace '<1' with '0.5' for numerical conversion
    df_filtered[col] = df_filtered[col].replace('<1', '0.5')
    # Replace '-' with '0' for numerical conversion
    df_filtered[col] = df_filtered[col].replace('-', '0')
    # Convert to numeric, coercing errors to NaN
    df_filtered[col] = pd.to_numeric(df_filtered[col], errors='coerce')
    # Fill any remaining NaNs (e.g., original missing in % Other or from coercion) with 0
    df_filtered[col] = df_filtered[col].fillna(0)

print("Filtered DataFrame head:")
print(df_filtered.head())
print(
"\nData types of diversity columns after conversion:")
print(df_filtered[diversity_cols].dtypes)
print("\nMissing values in filtered DataFrame:")
print(df_filtered[diversity_cols].isnull().sum())
mean_diversity_representation = df_filtered[diversity_cols].mean()

print("Average racial/ethnic representation (2014-2018):")
print(mean_diversity_representation)
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'Year' and calculate the mean of diversity columns
diversity_trends = df_filtered.groupby('Year')[diversity_cols].mean()

# Plotting the trends
plt.figure(figsize=(12, 7))
sns.lineplot(data=diversity_trends, markers=True)

plt.title('Racial/Ethnic Diversity Trends in Tech Companies (2014-2018)')
plt.xlabel('Year')
plt.ylabel('Average Percentage (%)')
plt.legend(title='Racial/Ethnic Group', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(diversity_trends.index) # Ensure all years are shown on the x-axis
plt.tight_layout()
plt.show()
```
